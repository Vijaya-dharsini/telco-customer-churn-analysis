# EDA Summary – Telco Customer Churn Analysis

* The dataset contains 7,043 customer records and 21 columns.
* No duplicate records were found, but 11 missing values were identified in the `TotalCharges` column.
* The overall customer churn rate is approximately 26.5%.
* Month-to-month contracts are the most common contract type.
* Fiber optic customers have the highest churn count among internet service categories.
* Customers who churned had shorter average tenure than customers who stayed.
* Customers who churned had higher average monthly charges than customers who stayed.

## Conclusion

Customer churn may be associated with contract type, tenure, internet service, and monthly charges. These findings will help guide the next stage of analysis and dashboard development.
